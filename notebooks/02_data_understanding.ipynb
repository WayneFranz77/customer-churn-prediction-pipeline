## Impact on Machine Learning

The distribution of the target variable can directly affect model performance.

### Balanced Datasets

When classes are relatively balanced:

- Accuracy can provide a useful measure of performance
- Models generally have sufficient examples of each outcome

### Imbalanced Datasets

When one class significantly outnumbers the other:

- Accuracy may become misleading
- Minority class predictions become more difficult
- Additional evaluation metrics become more important

---

## Evaluation Considerations

If the dataset is imbalanced, greater emphasis should be placed on metrics such as:

- Precision
- Recall
- F1 Score
- ROC-AUC

These metrics provide a more informative assessment of model performance than Accuracy alone, particularly when predicting customer churn.
