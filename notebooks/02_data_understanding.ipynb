TotalCharges Investigation
During the data type assessment, the TotalCharges variable was identified as a string (object) rather than a numerical variable.

Because TotalCharges represents a monetary amount, it should ideally be stored as a numeric feature.

The objective of this investigation is to determine whether the column contains:

Blank values
Non-numeric values
Formatting inconsistencies
Identifying these issues early will help ensure accurate preprocessing and model development.

```python
pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
).isnull().sum()
```

Findings
The investigation identified a data quality issue within the TotalCharges feature.

TotalCharges Conversion
When attempting to convert TotalCharges to a numerical data type, 11 values could not be converted successfully.

This indicates that 11 records contain non-numeric values, which are most likely blank strings rather than valid monetary amounts.

Why This Matters
TotalCharges is expected to represent the total amount paid by a customer and should therefore be stored as a numerical variable.

Machine learning models cannot directly use non-numeric values in a feature that represents monetary data.

Planned Treatment
During the data cleaning phase, the following steps will be performed:

Convert TotalCharges to a numeric variable.
Replace invalid values with missing values (NaN).
Determine an appropriate strategy for handling the 11 affected records.
Key Observation
This is the first confirmed data quality issue identified in the dataset and will be addressed during the data preparation and feature engineering stages.
