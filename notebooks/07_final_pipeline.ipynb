```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import GradientBoostingClassifier

from sklearn.metrics import (accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score)

print("Libraries Loaded")
```

Load and Prepare Dataset

The final pipeline will be trained using the cleaned dataset and the same preprocessing strategy established during feature engineering.

```python
df = pd.read_csv("../data/raw/Telco-Customer-Churn.csv")

df["TotalCharges"] = pd.to_numeric(df["TotalCharges"],errors="coerce")

df = df.dropna(subset=["TotalCharges"])

df = df.drop("customerID", axis=1)

X = df.drop("Churn", axis=1)
y = df["Churn"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42,
    stratify=y)

print("Dataset Ready")
```

```python
numerical_features = ["SeniorCitizen", "tenure", "MonthlyCharges", "TotalCharges"]

categorical_features = [col for col in X.columns if col not in numerical_features]
```

```python
preprocessor = ColumnTransformer(transformers=[("num", StandardScaler(), numerical_features),
    ("cat",OneHotEncoder(handle_unknown="ignore"), categorical_features)])
```

Build Final Candidate Models

Both Gradient Boosting and Logistic Regression will be evaluated one final time before selecting the deployment candidate.

```python
final_models = {"Logistic Regression": LogisticRegression(random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42)}
```

```python
final_results = []

for name, model in final_models.items():

    pipeline = Pipeline(steps=[("preprocessor", preprocessor), ("classifier", model)])

    pipeline.fit(X_train, y_train)

    y_pred = pipeline.predict(X_test)

    y_prob = pipeline.predict_proba(X_test)[:, 1]

    final_results.append({"Model": name, "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred,pos_label="Yes"),
        "Recall": recall_score(y_test, y_pred, pos_label="Yes"),
        "F1 Score": f1_score(y_test, y_pred, pos_label="Yes"),
        "ROC-AUC": roc_auc_score((y_test == "Yes").astype(int), y_prob)})

final_results_df = pd.DataFrame(final_results)

final_results_df
```

Final Model Selection

Two strong candidate models emerged from the model comparison phase:

Logistic Regression and Gradient Boosting

Although Gradient Boosting achieved the highest ROC-AUC score, Logistic Regression delivered:

Higher Accuracy
Higher Precision
Higher Recall
Higher F1 Score
The ROC-AUC difference between the two models was minimal.

Given its strong predictive performance, simplicity, interpretability, and deployment advantages, Logistic Regression was selected as the final production model.

Train Final Production Model

The selected Logistic Regression model will now be trained using the full production pipeline.

This pipeline combines:

Feature preprocessing
Feature encoding
Feature scaling
Classification
into a single reusable workflow.

```python
final_pipeline = Pipeline(steps=[("preprocessor", preprocessor),
        ("classifier", LogisticRegression(random_state=42))])

final_pipeline.fit(X_train, y_train)

print("Final Pipeline Trained")
```

Save Final Model

The final machine learning pipeline will be serialized and saved for future use.

This allows the model to be loaded later without requiring retraining.

```python
import joblib
```

```python
joblib.dump(final_pipeline, "../models/trained_models/customer_churn_pipeline.pkl")

print("Model Saved Successfully")
```
